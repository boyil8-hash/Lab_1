```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
plt.rcParams["figure.figsize"] = (15, 10)
```

```python
x=0.03
stats.norm.ppf(x, loc =0, scale = 1) #what's the position if we know the sigma, mean, and the area to the left) 
# x is area to the left, loc is mean, scale is sigma(sd) and this mataches with 
# z is (x-mu)/sigma which is (-1.8-0)/1 which is -1.88, and i check z table around -1.88, area is around 0.031 as what we expected 
# the minus signs appears because the z can be negative and when we choose the area to the left less than 0.5 for a standard normal distribution,the z is smaller than 0
# mathemtically z is (x-mu)/sigma, since x is negative when area is smaller than 0.5, mu is 0 and sigma is 1, it forces z to be negative 
```

```python
d = stats.expon.rvs(loc=2, scale=2, size=100000) # pull 100k random samples from a exponential distribution that is centered at 2 with sd of 2 
fig, ax = plt.subplots(1,1)  # 1 plot 
ax.hist( d,100, density=True ) # create a probability density histogram based on those samples 
plt.tick_params(labelsize =24) 
plt.xlim([1.0, 10.0])
x = np.linspace(1, 10.0,1000) # devide x from 1 to 10 into 1000 segements
ax.plot(x, stats.expon.pdf(x,loc = 2, scale = 2), linewidth =8, alpha =0.7)  # compute the value of y from 1000 segements of we chose using 0.5 e^ (-(x-2)/2),
# and plot the line 
plt.show() 

# choose 7 as the strength of the signal
# the question i want to ask is whether the signal ,that is equal to bigger than 7, is interesting or not 
y=7 
b= stats.expon.cdf(y, loc=2, scale=2) # One way we want to know whether a noise is interesting or not is to find the percentage 
# of noise that is equal to stronger than 7. we can write mathemtically as intergral of 0.5 e^ (-(x-2)/2) from 0 to 7 
a= 1.0 - b 
print (a) # this is the area to the right 

c = stats.norm.ppf(b, loc =0, scale = 1) # b is the area to the left, so let's use b here instead of a. what we get
#  from stats.norm.ppf(b, loc =0, scale = 1) is the corresonponding postition according to the standard norm distirbution 
print(c)

# part 4 
d= stats.norm.ppf(a, loc =0, scale = 1) # playing around changing the point to a which is area we were refering to 
e = stats.norm.cdf(d, loc =0, scale = 1) # playing around changing the the value of z, to see whether it gets a 
print(d)
print(e)
```
