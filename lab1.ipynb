```python
x=0.03
stats.norm.ppf(x, loc =0, scale = 1) 
# it returns the position z, if we give the sigma(sd), mean, and the area to the left
# x is area to the left, loc is mean, scale is sigma(sd) and this mataches with 
# area 0.03 is at about z = -1.88 acording to the z table as expected 
```

### D: why is the answer negative 
for p < 0.5, less than half area is to the left so z must lie below the mean,
and in this case z<0. The sign only tells us which side of the mean we are on.
mathemtically z is (x-mu)/sigma, since x is the position and it is negative when area is smaller than 0.5, mu is 0 and sigma is 1, it forces z to be negative.
Most of time in physics, "sigma" is quoted for the right tail, in this case, we can use " stats.norm.ppf(1-p)" to generate a postive sigma. 

```python
x=0.8413
stats.norm.ppf(x, loc =0, scale = 1) 
#this line above return the position if we give the sigma(sd), area to the left, and the position 
# area to the left of z =1 is 0.8413, so we expect z=1
```

### part 2: Expoential distribution

```python
d = stats.expon.rvs(loc=2, scale=2, size=100000) 
# it pulls 100k random samples from a exponential distribution that is begins at 2 with sd of 2 
fig, ax = plt.subplots(1,1)  # 1 plot 
ax.hist( d,100, density=True ) # create a probability density histogram based on those samples 
plt.tick_params(labelsize =24) 
plt.xlim([1.0, 10.0])
x = np.linspace(1, 10.0,1000) # devide x from 1 to 10 into 1000 segements
ax.plot(x, stats.expon.pdf(x,loc = 2, scale = 2), linewidth =8, alpha =0.7) 
# compute the value of y from 1000 segements of we chose using 0.5 e^ (-(x-2)/2),
# and plot the line 
plt.show() 
```

### Question: 
### if there is no signal, what's probability that background produce a measure that is  at least 7?
### or is the noise greater than 7 produced by the background interesting or not? 

### C-E: 

```python
y=7 
b= stats.expon.cdf(y, loc=2, scale=2) 
# One way we want to know whether a noise is interesting or not is to find
# the percentage of noise that is equal to stronger than 7. we can write
# mathemtically as INTEGRAL of 0.5 e^ (-(x-2)/2) from 7 to infinity 
a= 1.0 - b 
print (a) # this is the area to the right 

c = stats.norm.ppf(b, loc =0, scale = 1) 
# b is the area to the left, so let's use b here instead of a. what we get
# from stats.norm.ppf(b, loc =0, scale = 1) is the corresonponding postition according to the standard norm distirbution 
print(c)
```

```python
arrayNum = np. array ([3,4,5,7,10,15,20]) 
# several numbers for trials
b = stats.expon.cdf(arrayNum,loc=2, scale=2) 
a= 1.0 - b  
sigma = stats.norm.ppf(b, loc =0, scale = 1) 
print(arrayNum) 
print(a) 
print(sigma) 

# comments: 
# when y increases, probability that background produces drops quickly,
# but sigma grows much more slowly, so y =7 only gives 1.4 sigma. and you need 
# y = 15 to reach 3 sigma. This is because exponential have a longer tail than 
# standard normal distribution, so large value are more common than 
# standard normal distribution would predict. 
```
